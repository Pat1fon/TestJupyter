```python
import pandas as pd
import struct
import numpy as np
import os
```

```python
test_file_path = "D:\\TestJupyter\\data\\Stud_IT_2025_Module_2\\binary_data_train\\0\\img_000001_label_0.mybin"
with open(test_file_path, 'rb') as raw_data:
    read_data_head = raw_data.read(12)
    bin_data = raw_data.read()
```

```python
# 4 char 4 uint32 1 uint8 1 uint8 2 byte
unpacked_head = struct.unpack('4s i B B h', read_data_head)
unpacked_head
```

```python
#head_struct = struct.Struct('4s i B B h')
def unpacking_row(path, i):
    with open(path, 'rb') as raw_data:
        read_head = raw_data.read(12)
        read_bin_data = raw_data.read()
        #unpacked_head = head_struct.unpack(read_head)
        bin_data = np.frombuffer(read_bin_data, dtype = np.uint8)
        row = {
            'Label' : i,
            #'Type' : unpacked_head[0],
            #'Amount' : unpacked_head[1],
            #'Version' : unpacked_head[2],
            #'Encoding' : unpacked_head[3],
            'Data' : bin_data
        }
        return row
        
```

```python
unpacking_row(test_file_path, 0)
```

```python
path = "D:\\TestJupyter\\data\\Stud_IT_2025_Module_2\\binary_data_train"
```

```python
data_list = []
for i in range(0, 10):
    dir_path = path+f'\\{i}'
    file_names = os.listdir(dir_path)
    for file in file_names:
        file_path = dir_path + '\\' + file
        row = unpacking_row(file_path, i)
        data_list.append(row)
df = pd.DataFrame(data_list)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
def count_plot(column, title):
    plt.figure(figsize=(10, 6))
    sns.countplot(x=column, data = df, hue=column)
    plt.title(title)
    plt.show()
```

```python
X = np.stack(df['Data'])
y = df['Label']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
```

```python
pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('classifier', None)
])
```

```python

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = [
    {
        'classifier': [
            RandomForestClassifier(
            random_state=42, 
            class_weight='balanced'
        )],
        'classifier__n_estimators': [50, 100, 200, 400],
        'classifier__max_depth': [None, 5, 10, 20]
    },
    {
        'classifier': [DecisionTreeClassifier(random_state=42)],
        'classifier__criterion': ['gini', 'entropy', 'log_loss'],
        'classifier__splitter': ['best', 'random'],
        'classifier__max_depth': [None, 5, 10],
        'classifier__max_features': [ None, 'sqrt', 'log2']
    }
]
```

```python
param_grid = [
    {
        'classifier': [RandomForestClassifier(random_state=42, class_weight='balanced', n_jobs=-1)],
        'classifier__n_estimators': [50, 100, 200, 400],
          # Проверьте сначала одно дефолтное значение
        'classifier__max_depth': [20]
    },
]
```

```python
grid = GridSearchCV(pipe, param_grid, cv=5, scoring='accuracy', n_jobs=-1)
```

```python
grid.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
print(f"Победитель: {grid.best_params_}")
print(f"Лучший средний CV-score: {grid.best_score_:.4f}")
best_pipeline = grid.best_estimator_
final_preds = best_pipeline.predict(X_test)
print(f"Финальный Accuracy на тесте: {accuracy_score(y_test, final_preds):.4f} /n {classification_report(y_test, final_preds)}")
```

```python
model = best_pipeline
```
